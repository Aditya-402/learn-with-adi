## 8.2 · The everyday methods — and the rule: methods return NEW strings, they never edit

## 8.3 · Split, join & search — the revolving door between text and lists

```python
line = "Ada,36,London"
parts = line.split(",")
print(parts)
print(f"{parts[0]} lives in {parts[2]}")
print(" | ".join(parts))
```

```python
# P3 — count & find. Predict both numbers before running.
quote = "to be or not to be"
print(quote.count("to"))
print(quote.find("not"))
```

## 8.4 · Formatting, the deep dive — width lanes (`:<10`, `:>7.2f`) and `\n`

```python
print(f"{'Item':<10}{'Price':>7}")
print("-" * 17)
print(f"{'Tea':<10}{12.5:>7.2f}")
print(f"{'Samosa':<10}{45:>7.2f}")
print(f"{'Biscuit':<10}{8:>7.2f}")
print("-" * 17)
total = 12.5 + 45 + 8
print(f"{'Total':<10}{total:>7.2f}")
```
