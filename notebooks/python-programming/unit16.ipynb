## 16.1 · The open-door problem — and the underscore handshake

```python
class Jar:
    def __init__(self, cookies=0):
        self._cookies = cookies      # the sign: internal

    def count(self):                 # the polite way in
        return self._cookies

j = Jar(12)
print("Cookies:", j.count())

j._cookies = 999                     # allowed! nothing stops you
print("Cookies:", j.count())         # ...and the sign was ignored
```

## 16.2 · @property — a guarded window (read with NO parentheses)

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance      # the real box, behind the sign

    @property
    def balance(self):               # the window
        return self._balance

class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):                  # computed, never stored
        return 3.14159 * self.radius ** 2

a = Account("Ada", 500)
print("Balance:", a.balance)         # no parentheses!

c = Circle(2)
print(f"Area at r=2: {c.area:.2f}")
c.radius = 3
print(f"Area at r=3: {c.area:.2f}")  # fresh, automatically
```

```python
# P2 — run as-is and read the LAST line of the report. A read-only property is enforced.
class Book:
    def __init__(self, pages):
        self._pages = pages

    @property
    def pages(self):
        return self._pages

b = Book(320)
print(b.pages)
b.pages = 500      # the window doesn't open
print(b.pages)     # this line never runs
```

## 16.3 · Setters that say no — validate at the boundary, trust the inside

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):               # the getter — must come first
        return self._balance

    @balance.setter
    def balance(self, value):        # the setter — same name, on purpose
        if value < 0:
            raise ValueError("balance cannot be negative")
        self._balance = value

a = Account("Ada", 500)
a.balance = 800                      # runs the setter, passes the check
print(a.balance)

try:
    a.balance = -999                 # runs the setter, hits the raise
except ValueError as e:
    print("Refused:", e)

print(a.balance)                     # untouched — the guard held
```

## 16.4 · Class attributes vs instance attributes — lookup goes up, assignment stays put

```python
class Account:
    bank_name = "Learn Bank"     # shared by every instance

    def __init__(self, owner):
        self.owner = owner       # one per instance

a = Account("Ada")
b = Account("Grace")
print(a.bank_name, "|", b.bank_name)
print(Account.bank_name)         # reachable on the class too

Account.bank_name = "Learn Bank Ltd"   # one edit, on the shelf
print(a.bank_name, "|", b.bank_name)   # both objects follow
```

## 16.5 · Your build — BankAccount, final form (the Intermediate checkpoint)

`_balance` hidden · `balance` read-only · `deposit`/`withdraw` validating · class attribute `bank_name` in `report()` · `__str__` from Unit 14. Then a scripted scenario that walks into every guard.

```python
class BankAccount:
    bank_name = "Learn Bank"

    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance

    # balance property (read-only)
    # deposit / withdraw with their guards
    # report()  ->  "Learn Bank | Ada: 1200.00"
    # __str__   ->  "BankAccount(Ada, 1000)"


a = BankAccount("Ada", 1000)
print(a)

for amount in [500, -50]:
    try:
        a.deposit(amount)
        print("Deposited", amount)
    except ValueError as e:
        print("Refused:", e)

for amount in [300, 9000]:
    try:
        a.withdraw(amount)
        print("Withdrew", amount)
    except ValueError as e:
        print("Refused:", e)

try:
    a.balance = 999999
except AttributeError:
    print("Refused: balance is read-only")

print(a.report())
```
