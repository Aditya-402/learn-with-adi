## 10.1 · Writing files — a box that outlives the program

`open(name, "w")` hands you a blank page; `write()` puts down *exactly* what you give it — the `
` is yours to add. The `with` block closes the file for you, even if an error strikes.

```python
# the old way, then the better way — and watch "w" wipe
f = open("notes.txt", "w")
f.write("First note
")
f.close()

with open("notes.txt", "w") as f:   # "w" again: blank page, first note gone
    f.write("Second note
")

print(open("notes.txt").read().strip())
```

## 10.2 · Reading it back — and the invisible `
` returns

`read()` gives one big string, `readlines()` a list, and `for line in f:` loops line by line. Every line keeps its trailing `
` — `strip()` it (Unit 8's fix, back again).

```python
with open("menu.txt", "w") as f:
    f.write("idli
dosa
vada
")

with open("menu.txt") as f:
    lines = f.readlines()
print(lines)                    # see the 
 stowaways

with open("menu.txt") as f:
    for line in f:
        print(line.strip())     # the clean read
```

## 10.3 · Appending & guarding — mode `"a"` and the file that isn't there

`"a"` appends (and creates the file if missing). Reading a missing file raises `FileNotFoundError` — real apps catch it and treat it as *first run, start fresh*.

```python
with open("log.txt", "a") as f:     # session 1 (creates the file)
    f.write("visit 1
")

with open("log.txt", "a") as f:     # session 2 — pencil starts at the end
    f.write("visit 2
")

with open("log.txt") as f:
    for line in f:
        print(line.strip())

# run this cell again — in Colab the log REALLY grows
```

## 10.4 · Structured data — json, the round trip

`json.dump(value, f)` writes a list/dict as agreed-upon text; `json.load(f)` rebuilds it, **types intact**. `f.write(str(tasks))` only writes a photo of your list — it comes back as one string.

```python
import json

tasks = ["buy milk", "call grandma"]
with open("todo.json", "w") as f:
    json.dump(tasks, f)

print("raw file:", open("todo.json").read())

with open("todo.json") as f:
    loaded = json.load(f)
print("loaded:  ", loaded)
print("type:", type(loaded))
print("first task:", loaded[0])
```

## 10.5 · Your build — the to-do list that survives a restart

On the study-notes page the two "runs" are folded into one program. Here they don't have to be: run the cell below, **restart the runtime** (Runtime → Restart), then run it again — your tasks come back from `todo.json`. That's the whole point of this unit.

```python
import json

def load_tasks():
    try:
        with open("todo.json") as f:
            return json.load(f)
    except FileNotFoundError:
        return []

def save_tasks(tasks):
    with open("todo.json", "w") as f:
        json.dump(tasks, f)

def add_task(tasks, item):
    tasks.append(item)

def show_tasks(tasks):
    n = 1
    for t in tasks:
        print(f"{n}. {t}")
        n = n + 1

tasks = load_tasks()               # finds yesterday's file — or starts fresh
add_task(tasks, input("Add a task: "))
save_tasks(tasks)
show_tasks(tasks)
```
