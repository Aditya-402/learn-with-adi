## 3.3 · elif — many branches, first True wins (order matters!)

```python
marks = 78
if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 50:
    grade = "C"
else:
    grade = "D"
print(f"Grade: {grade}")
```

```python
# P2 — the upside-down ladder: a 95 gets a C! Reorder the questions (don't change numbers).
marks = 95
if marks >= 50:
    print("Grade: C")
elif marks >= 75:
    print("Grade: B")
elif marks >= 90:
    print("Grade: A")
else:
    print("Grade: D")
```
