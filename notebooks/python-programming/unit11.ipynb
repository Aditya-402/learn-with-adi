## 11.1 · import & math — your library card

`import math` fetches the shelf once and gives you ONE name, `math`; the dot reaches each tool: `math.sqrt`, `math.pi`, `math.floor`, `math.ceil`.

```python
import math

area = 240
side = math.sqrt(area)
print(f"Each side: {side:.2f} m")
print(f"Between {math.floor(side)} and {math.ceil(side)} whole metres")
```

## 11.2 · random — dice, draws & the replay switch

`randint(1, 6)` includes BOTH ends. `choice` picks from a list; `shuffle` rearranges it in place. And `random.seed(42)` pins the number recipe's start, so every run replays the same "random" stream — the reproducibility trick ML experiments live by.

```python
import random

menu = ["dosa", "noodles", "rajma rice", "sandwich"]
lunch = random.choice(menu)
payer = random.randint(1, 6)
print(f"Today's lunch: {lunch}")
print(f"Die says player {payer} pays!")
```

```python
# P3 — the replay experiment. Run this cell three times: same three rolls.
# Change 42 to 7: a different trio, but again pinned. Delete the seed line: chaos.
import random

random.seed(42)
print(random.randint(1, 6))
print(random.randint(1, 6))
print(random.randint(1, 6))
```

## 11.3 · datetime — the calendar shelf

`from datetime import date` (the from-import earning its keep), `date(2026, 8, 15)` builds a date, subtracting two dates gives a gap whose `.days` is the whole-day count — leap years handled.

```python
from datetime import date

today = date.today()   # a real clock, here in Colab
newyear = date(today.year + 1, 1, 1)
gap = newyear - today
print(f"Today is {today}")
print(f"{gap.days} days until {newyear.year}!")
```

## 11.5 · Your build — the dice duel

Seeded with 42 so the duel replays identically (that's how the page checks it). Then delete the seed line and run five times — the game only becomes a game once the replay switch comes out.

```python
import random

random.seed(42)
p1a = random.randint(1, 6)
p1b = random.randint(1, 6)
p2a = random.randint(1, 6)
p2b = random.randint(1, 6)
p1 = p1a + p1b
p2 = p2a + p2b
print(f"Player 1 rolls {p1a} and {p1b} (total {p1})")
print(f"Player 2 rolls {p2a} and {p2b} (total {p2})")
if p1 > p2:
    print("Player 1 wins!")
elif p2 > p1:
    print("Player 2 wins!")
else:
    print("It's a tie!")
```
