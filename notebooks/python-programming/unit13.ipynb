## 13.3 · Methods & self — abilities that know whose they are

```python
class Player:
    def __init__(self, name, health):
        self.name = name
        self.health = health

    def take_damage(self, amount):
        self.health = self.health - amount
        print(f"{self.name} takes {amount}, now on {self.health}")

p1 = Player("Ada", 100)
p2 = Player("Alan", 100)

p1.take_damage(30)
p2.take_damage(10)
p1.take_damage(25)
```

## 13.4 · Methods that answer vs methods that act

```python
class Player:
    def __init__(self, name, health):
        self.name = name
        self.health = health

    def take_damage(self, amount):    # acts
        self.health = self.health - amount

    def heal(self, amount):           # acts
        self.health = self.health + amount

    def is_alive(self):               # answers
        return self.health > 0

p = Player("Ada", 30)
p.take_damage(40)
print(p.is_alive())

if not p.is_alive():
    p.heal(50)

print(p.health, p.is_alive())
```

```python
# P2 — predict all five lines BEFORE running. Especially the last one.
class Square:
    def __init__(self, side):
        self.side = side

    def area_print(self):
        print(self.side * self.side)

    def area_return(self):
        return self.side * self.side

s = Square(4)
s.area_print()
print(s.area_return())
total = s.area_return() + 10
print(total)
oops = s.area_print()
print(oops)
```
