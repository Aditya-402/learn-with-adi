## 9.1 · Reading a traceback, properly

Run the crashing bill-splitter and read the report against the three moves: **last line** (the what) → **bottom `File` line** (the where) → **walk up** (the how).

```python
def share_per_person(total, people):
    return total / people

def print_share(total, people):
    each = share_per_person(total, people)
    print("Each pays:", each)

print_share("900", 3)   # run as-is; read the traceback bottom-up; then fix AT THE DOOR
```

## 9.4 · Debugging by printing — and raising your own

`print(f"DEBUG: {box=}")` shows the true state of memory mid-flight — the hand-trace, automated. And `raise ValueError("...")` enforces *your* rules with the same machinery Python uses for its own.

```python
minutes = 200
hours = minutes // 60
left = minutes % 60
print(f"DEBUG: {hours=} {left=}")
print(f"{hours} h {left} min")
```

## 9.5 · Your build — bulletproof the to-do app

Unit 7's function version, armour-plated three ways: unknown menu choice → friendly message; removing a missing task → friendly message (`.remove()` raises ValueError — a rogue you know); empty task name → polite refusal. Here the menu loop genuinely waits for your keyboard — be as rude a user as you can.

```python
def add_task(tasks, name):
    # refuse an empty name with "A task needs a name."
    tasks.append(name)
    print("Added:", name)

def remove_task(tasks, name):
    # wrap in try / except ValueError -> "Not on the list: <name>"
    tasks.remove(name)
    print("Removed:", name)

def show_tasks(tasks):
    for t in tasks:
        print("-", t)

tasks = []
while True:
    choice = input("add / remove / show / quit? ")
    if choice == "add":
        add_task(tasks, input("Task name? "))
    elif choice == "remove":
        remove_task(tasks, input("Which task? "))
    elif choice == "show":
        show_tasks(tasks)
    elif choice == "quit":
        print("Bye!")
        break
    # answer unknown choices with a friendly message
```
