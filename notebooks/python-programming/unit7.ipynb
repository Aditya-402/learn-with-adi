## 7.2 · Ingredients — parameters (slots on the card) vs arguments (what you pass in)

```python
def greet(name, punctuation="!"):
    print(f"Hello, {name}{punctuation}")

greet("Ada")          # leans on the default
greet("Rhea", "?")    # overrides it
greet("Sam", "...")
```

## 7.3 · return vs print — print shouts and forgets; return hands the value back

```python
def area(width, height):
    return width * height

hall = area(9, 5)
kitchen = area(3, 4)
total = hall + kitchen
print(f"Hall: {hall}")
print(f"Kitchen: {kitchen}")
print(f"Total: {total}")
```

## 7.4 · Scope, gently — boxes made inside a function vanish when the call ends

```python
snack = "biscuit"

def picnic():
    snack = "mango"          # a bench box that SHADOWS the outer one
    print("Inside:", snack)

picnic()
print("Outside:", snack)      # the room's box was untouched
```

```python
# P1 — predict, then run: what happens on the last line, and why?
def bake():
    temperature = 180
    print("Baking at", temperature)

bake()
print(temperature)
```

```python
# Build (a) — the Unit 5 version, working but tangled. Refactor me into three functions.
tasks = []
running = True
while running:
    command = input("Command? ")
    if command == "add":
        name = input("Task? ")
        tasks.append(name)
        print(f"Added: {name}")
    elif command == "show":
        if len(tasks) == 0:
            print("Nothing to do!")
        else:
            n = 1
            for task in tasks:
                print(f"{n}. {task}")
                n = n + 1
    elif command == "remove":
        name = input("Task? ")
        if name in tasks:
            tasks.remove(name)
            print(f"Removed: {name}")
        else:
            print(f"Not found: {name}")
    elif command == "quit":
        running = False
        print("Bye!")
```
