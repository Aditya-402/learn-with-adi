## 21.2 · Vectorised arithmetic — the operation reaches every element at once

```python
import numpy as np

a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print(a + b)     # element-wise
print(a - b)
print(a * b)
print(a / b)     # floats, always
print(a + 5)     # the 5 reaches every element
print(a > 25)    # a question asked of all four
```

## 21.3 · Shape — say it out loud at every step

```python
import numpy as np

a = np.array([10, 20, 30, 40])
b = np.array([[1, 2, 3],
              [4, 5, 6]])

print(a.shape, a.ndim)
print(b.shape, b.ndim)
print(b.reshape(3, 2))     # same six numbers, regrouped
print(b.reshape(6))        # flattened back to one direction
print(b.reshape(6).shape)
```

```python
# P3 (stretch) — a row is not a column. (3,) is NOT (3, 1).
v = np.array([1, 2, 3])
col = v.reshape(3, 1)
print(v.shape)
print(col.shape)
print(col)
```

## 21.4 · Indexing, slicing & boolean masks — a[:, 0] is a whole column

```python
import numpy as np

grid = np.array([[10, 20, 30],
                 [40, 50, 60],
                 [70, 80, 90]])

print(grid[1, 2])         # one cell: row 1, column 2
print(grid[0])            # a whole row
print(grid[:, 0])         # a whole column
print(grid[0:2, 1:3])     # a rectangular block
print(grid[grid > 45])    # a mask — flat result
print(grid[:, 0].shape, grid[grid > 45].shape)
```

## 21.5 · Aggregations & axis — axis is the direction that disappears

```python
import numpy as np

t = np.array([[2, 4, 4, 4],
              [5, 5, 7, 9]])

print(t.sum(), t.mean(), t.min(), t.max())
print(f"Spread: {t.std():.2f}")

print(t.sum(axis=0))        # down the columns → one per column
print(t.sum(axis=1))        # across the rows  → one per row
print(t.sum(axis=0).shape, t.sum(axis=1).shape)
```

## 21.6 · Your build — the marks report, NumPy edition

Unit 12 built this report with a loop over `zip()` and two comprehensions. Build it again out of one 2-D array and count the loops: there should be none.

```python
import numpy as np

names = ["Ada", "Bob", "Cara", "Dev"]
subjects = ["Maths", "Physics", "Python"]
marks = np.array([[91, 78, 85],
                  [45, 32, 58],
                  [78, 88, 92],
                  [85, 70, 76]])

student_avg = marks.mean(axis=1)
subject_avg = marks.mean(axis=0)
top = student_avg.argmax()

print("=== MARKS REPORT (NumPy) ===")
print(f"Shape: {marks.shape}")
print("-- Per student --")
print(f"{names[0]:<8}{student_avg[0]:.2f}")
print(f"{names[1]:<8}{student_avg[1]:.2f}")
print(f"{names[2]:<8}{student_avg[2]:.2f}")
print(f"{names[3]:<8}{student_avg[3]:.2f}")
print("-- Per subject --")
print(f"{subjects[0]:<8}{subject_avg[0]:.2f}")
print(f"{subjects[1]:<8}{subject_avg[1]:.2f}")
print(f"{subjects[2]:<8}{subject_avg[2]:.2f}")
print(f"Class average: {marks.mean():.2f}")
print(f"Passed: {(student_avg >= 50).sum()} of {student_avg.size}")
print(f"Top scorer: {names[top]} ({student_avg[top]:.2f})")
```
