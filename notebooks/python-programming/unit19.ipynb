## 19.1 · The function IS a value — `greet` is the recipe card, `greet()` is the cooking

```python
# One card, two labels — and a helper that takes a card as an argument.
def apply_twice(func, value):
    return func(func(value))

def add_five(n):
    return n + 5

def shout(text):
    return text.upper() + "!"

print(apply_twice(add_five, 10))
print(apply_twice(shout, "hi"))
```

## 19.2 · lambda and the `key=` trick — the most common real-world lambda sighting

```python
students = [
    {"name": "Ada", "marks": 78},
    {"name": "Ravi", "marks": 91},
    {"name": "Mira", "marks": 84},
]

for s in sorted(students, key=lambda s: s["marks"], reverse=True):
    print(f"{s['name']}: {s['marks']}")

top = max(students, key=lambda s: s["marks"])
print(f"Top: {top['name']}")
```

## 19.3 · Closures — the function factory, and the packed lunchbox

```python
def make_multiplier(n):
    return lambda x: x * n

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(5))
print(triple(5))
print(double(triple(2)))   # each carries its own n
```

## 19.4 · `*args` and `**kwargs` — functions that accept anything (and the wrapper shape)

```python
def summary(*args):
    print(f"got {len(args)} numbers, total {sum(args)}")

summary(3, 5, 9)
summary(10)
summary()

def profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

profile(name="Ada", city="Chennai", marks=91)

# The honest purpose: wrappers that pass everything through.
# Remember this shape — Unit 20 is built on it.
def shout(text, times=1):
    return (text.upper() + "! ") * times

def log_call(func, *args, **kwargs):
    print("calling", func.__name__)
    return func(*args, **kwargs)

print(log_call(shout, "ready", times=2))
```

## 19.5 · Your build — the leaderboard (tuple key) and the event system (functions in a dict)

```python
# Score descending, ties broken by name ascending — one tuple key does both.
players = [
    {"name": "Mira", "score": 90},
    {"name": "Ada", "score": 120},
    {"name": "Ravi", "score": 90},
    {"name": "Bhavana", "score": 145},
]

ranked = sorted(players, key=lambda p: (-p["score"], p["name"]))
for rank, p in enumerate(ranked, start=1):
    print(f"{rank}. {p['name']} ({p['score']})")
```

```python
# The event system: a dict of event name -> handler card. .get keeps 'fly' polite.
def on_start():
    print("Engine started.")

def on_stop():
    print("Engine stopped.")

handlers = {
    "start": on_start,
    "stop": on_stop,
    "beep": lambda: print("Beep!"),
}

def fire(event):
    handler = handlers.get(event)
    if handler is None:
        print(f"No handler for '{event}' (ignored).")
    else:
        handler()

for e in ["start", "beep", "fly", "stop"]:
    fire(e)
```
