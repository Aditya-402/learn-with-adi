## 6.1 · Labels instead of positions — the dictionary

`d[key]` crashes on a missing key (read the last line first); `d.get(key, fallback)` stays calm.

```python
phone = {"Amma": "98450 11111", "Ravi": "72004 22222"}
print("Ravi:", phone["Ravi"])
print("Zara:", phone.get("Zara", "not saved yet"))
```

## 6.2 · Growing and changing — a dict is alive

`d[key] = value` creates the slot **or** replaces its contents — same syntax, no warning. Plus `del d[key]`, `key in d`, `len(d)`.

```python
cart = {"soap": 2}
cart["soap"] = cart["soap"] + 1   # replace: one more soap
cart["shampoo"] = 1               # create: brand-new slot, same syntax
del cart["soap"]                  # changed our mind entirely
print("Different items:", len(cart))
print("Soap still there?", "soap" in cart)
```

## 6.3 · Looping a dict — visiting every pair

`for key in d:` — tags only · `d.values()` — contents only · `for key, value in d.items():` — both at once, in insertion order.

```python
menu = {"tea": 12, "coffee": 15, "cake": 40}
total = 0
for item, price in menu.items():
    print(f"{item}: {price}")
    total = total + price
print(f"Everything once: {total}")
```

## 6.4 · Tuples & sets, briefly

A tuple is frozen at birth — `(3, 4)` — read it, unpack it (`x, y = point`), never edit it. A set keeps one copy of each value and promises **no order** — so always print `sorted(the_set)`, never the raw set.

```python
point = (3, 4)
x, y = point
print("x + y =", x + y)

visitors = ["adi", "meera", "adi", "ravi", "meera", "adi"]
distinct = set(visitors)
print("Distinct visitors:", len(distinct))
print(sorted(distinct))
```

## 6.5 · Your build — contact book & word counter

Here `input()` genuinely waits — add whoever you like. The word counter's `counts.get(word, 0) + 1` line is the dict-accumulator pattern; it returns, barely disguised, in the ML stage.

```python
# Contact book v1 — two saved contacts, one new from the keyboard
contacts = {"Amma": "98450 11111", "Ravi": "72004 22222"}
name = input("New contact name? ")
number = input("Their number? ")
contacts[name] = number
print("Ravi's number:", contacts["Ravi"])
print("All contacts:")
for person, num in contacts.items():
    print(f"{person}: {num}")
```

```python
# The word counter — finish it: print each pair as  word: n
sentence = "the cat sat on the mat the end"
counts = {}
for word in sentence.split():
    counts[word] = counts.get(word, 0) + 1
```
