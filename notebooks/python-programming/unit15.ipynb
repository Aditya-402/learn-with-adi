## 15.1 · Is-a: inheriting — a class built on another class

`class SavingsAccount(BankAccount):` means "same as the house dal, plus…". The child gets every method free; Python looks in the child first, then climbs to the parent.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance = self.balance - amount

    def report(self):
        print(f"{self.owner}: {self.balance:.2f}")


class SavingsAccount(BankAccount):
    pass


s = SavingsAccount("Meera", 500)
s.deposit(250)
s.withdraw(100)
s.report()
print(isinstance(s, SavingsAccount), isinstance(s, BankAccount))
```

## 15.2 · Extending with super() — "same as the parent, plus…"

`super()` means **my parent**. Run the parent's setup first, then add your own boxes. Forget it and you get `AttributeError: object has no attribute 'owner'` from somewhere innocent-looking.

```python
class SavingsAccount(BankAccount):
    def __init__(self, owner, balance, rate):
        super().__init__(owner, balance)
        self.rate = rate

    def add_interest(self):
        self.balance = self.balance + self.balance * self.rate / 100


s = SavingsAccount("Meera", 1000, 5)
s.add_interest()
s.report()
print(f"Rate: {s.rate}%")
```

## 15.3 · Overriding & polymorphism — in plain words

Same method name in the child → the child's version wins. Then one loop over mixed objects, each answering its own way. **That loop is polymorphism** — there is nothing more to the word.

```python
class SavingsAccount(BankAccount):
    def __init__(self, owner, balance, rate):
        super().__init__(owner, balance)
        self.rate = rate

    def report(self):
        print(f"{self.owner}: {self.balance:.2f} (savings at {self.rate}%)")


class PremiumAccount(BankAccount):
    def withdraw(self, amount):
        if amount > self.balance + 500:
            print("Over even the premium limit")
        else:
            self.balance = self.balance - amount

    def report(self):
        print(f"{self.owner}: {self.balance:.2f} [premium]")


account = BankAccount("Ravi", 1000)
savings = SavingsAccount("Meera", 2000, 5)
premium = PremiumAccount("Zoya", 300)

premium.withdraw(700)     # allowed — overdraft up to 500
account.withdraw(5000)    # refused by the parent's rule

for a in [account, savings, premium]:
    a.report()
```

## 15.4 · Has-a: composition — objects inside objects

An object is a box of boxes — and the inner boxes can hold **other objects**. Say "is-a" out loud: *a Library is a Book* is nonsense, so composition. *A SavingsAccount is a BankAccount* rings true, so inheritance. When in doubt, prefer composition.

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"{self.title} by {self.author}"


class Library:
    def __init__(self, name):
        self.name = name
        self.books = []

    def add(self, book):
        self.books.append(book)

    def show(self):
        print(f"{self.name} holds {len(self.books)} books:")
        for b in self.books:
            print(f"- {b}")


lib = Library("Adi's shelf")
lib.add(Book("Small Things", "Roy"))
lib.add(Book("Interpreter of Maladies", "Lahiri"))
lib.show()
```
