## 18.1 · yield — a function that pauses

Calling a generator function runs **nothing**. It builds the machine. Each `next()` runs to the next `yield` and freezes there.

```python
# Unit 17's whole Countdown class, in three lines of body.
def countdown(n):
    while n > 0:
        yield n
        n -= 1

print("before the loop")
for x in countdown(3):
    print(x)
print("done")
```

## 18.2 · Laziness — values on demand

A list is a warehouse; a generator is a kitchen that cooks to order. `range()` has been lazy since Unit 4 — you've been using this all along.

An infinite loop is safe here because `yield` stops it after every single value. The danger lives in whoever asks.

```python
def naturals():
    n = 1
    while True:
        yield n
        n += 1

taken = 0
for value in naturals():
    print(value)
    taken += 1
    if taken == 5:
        break

print("stopped at 5 — the machine is still frozen, waiting")
```

## 18.3 · Generator expressions

`[x*x for x in nums]` builds a list. `(x*x for x in nums)` builds a generator. Same shape, one character, enormous difference.

Never `print()` a generator itself — you get the box, not the contents.

```python
nums = [1, 2, 3, 4, 5]

print([x*x for x in nums])                 # eager: a real list
print(sum([x*x for x in nums]))            # builds a list, sums it, bins it
print(sum(x*x for x in nums))              # lazy: nothing but a running total
print(max(len(w) for w in ["fig", "banana", "kiwi"]))
```

## 18.4 · Pipelines — generators feeding generators

Build the chain and nothing happens. The last loop is the tap: it pulls one line through every stage, then the next. Memory stays flat however big the file is.

```python
with open("app.log", "w") as f:
    f.write("INFO  started\n")
    f.write("ERROR disk full\n")
    f.write("INFO  retry\n")
    f.write("ERROR timeout\n")

with open("app.log") as f:
    lines = (l.strip() for l in f)
    errors = (l for l in lines if l.startswith("ERROR"))
    numbered = enumerate(errors, 1)
    print("pipeline built — nothing has been read yet")
    for i, line in numbered:
        print(i, line)
```
