## 4.1 · while — repeat while true

Three sacred parts: the **start** (a box before the loop), the **condition** (a comparison that reads it), the **update line** (the line inside that changes it). Forget the update and the loop runs forever.

```python
count = 1
while count <= 5:
    print("Hello number", count)
    count = count + 1
print("Five hellos, four lines. Unit 2 forgiven.")
```

```python
# P3 — the deliberate infinite loop. Run it, watch it spin, press the STOP
# button (left of the cell). Then change something so the loop can end.
n = 1
while n > 0:
    n = n + 1
print("escaped")
```

## 4.2 · for + range — counted repetition

`for` is a `while` with the counting done for you. `range(5)` → 0..4 · `range(2, 6)` → 2..5 · `range(1, 10, 2)` → odd numbers. Range counts from 0 and always stops one short of the stop value.

```python
for i in range(1, 6):
    print(f"{i} squared is {i * i}")
```

```python
# P3 — print the even numbers 2 to 10, one per line, with NO if.
# The step argument does all the work. Mind the stop value!
for n in range(2, 3, 2):   # fix the range
    print(n)
```

## 4.3 · break & continue — leaving early, skipping one

`break` leaves the whole loop immediately; `continue` abandons only the current pass. And `while True:` + `break` is the honest shape for "loop until something happens mid-body".

```python
for n in range(101, 200):
    if n % 7 == 0:
        print("First multiple of 7 past 100:", n)
        break
```

```python
# P2 — predict the exact output BEFORE running: which values survive,
# and does the loop end early or run its course?
for n in range(1, 8):
    if n % 3 == 0:
        continue
    print(n)
print("all done")
```

## 4.4 · Counters & accumulators — loops that build an answer

The jar pattern: init **before** the loop (`total = 0`), update **inside** (`total = total + x`), read **after**. A counter is the same jar fed `+ 1`, often under an `if` — that's count-matching.

```python
total = 0
passed = 0
for i in range(4):
    score = int(input("Score? "))
    total = total + score
    if score >= 50:
        passed = passed + 1
print("Total:", total)
print("Passed:", passed)
```
