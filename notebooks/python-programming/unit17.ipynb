```python
nums = [1, 2, 3]
print("list, pass 1:", list(nums))
print("list, pass 2:", list(nums))

it = iter(nums)
print("iterator, pass 1:", list(it))
print("iterator, pass 2:", list(it))   # silence — the bookmark is spent
```

## 17.3 · `enumerate` & `zip` demystified — both are lazy iterators

```python
menu = ["idli", "dosa", "vada"]
print(list(enumerate(menu)))              # look inside with list()
print(list(enumerate(menu, start=1)))     # the 1-based numbering you used to hand-build
for n, item in enumerate(menu, start=1):
    print(f"{n}. {item}")

names = ["Ada", "Ben"]
scores = [90, 85]
for name, score in zip(names, scores):
    print(name, "scored", score)
```

## 17.4 · Writing your own — the Countdown class

Two dunders and a signal, and your class joins Python's own loop machinery.

```python
class Countdown:
    def __init__(self, start):
        self.n = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.n <= 0:
            raise StopIteration
        value = self.n
        self.n -= 1
        return value

for n in Countdown(3):
    print(n)
print("Liftoff!")
```

```python
# P3 — your class is one-shot too. Predict both lines before running.
c = Countdown(3)
print(list(c))
print(list(c))
```
