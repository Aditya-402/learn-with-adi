## 20.1 · Wrapping a function by hand — only Unit 19 tools

```python
def announce(func):
    def inner(*args, **kwargs):
        print("-- starting --")
        result = func(*args, **kwargs)
        print("-- finished --")
        return result
    return inner

def add(a, b):
    return a + b

loud_add = announce(add)      # runs NOW; builds a new function
print(loud_add(2, 3))         # inner opens its lunchbox and finds the original add
```

## 20.2 · The @ syntax — @announce above a def IS  add = announce(add)

```python
def logged(func):
    def inner(*args, **kwargs):
        print(f"calling {func.__name__} with {args}")
        result = func(*args, **kwargs)
        print(f"{func.__name__} returned {result}")
        return result
    return inner

@logged
def area(w, h):
    return w * h

print(area(3, 4))
```

```python
# P2 — predict all six lines BEFORE running. When does "decorating now" appear, and how often?
def noisy(func):
    print("decorating now")
    def inner(*args, **kwargs):
        print("running now")
        return func(*args, **kwargs)
    return inner

@noisy
def hello():
    print("hello")

print("--- defined ---")
hello()
hello()
```

## 20.3 · Decorators that count and transform

```python
def call_counter(func):
    def inner(*args, **kwargs):
        inner.count = inner.count + 1
        print(f"call #{inner.count}")
        return func(*args, **kwargs)
    inner.count = 0          # a box hung on the function object; it survives between calls
    return inner

@call_counter
def ping():
    print("pong")

ping()
ping()
ping()
```

```python
# P1 — write `excited`: uppercase the wrapped function's RETURN value and add "!"
def excited(func):
    pass

@excited
def greet(name):
    return f"hello, {name}"

print(greet("adi"))       # target ->  HELLO, ADI!
```

## 20.4 · Context managers — what `with` really does (Unit 10's promise, paid)

```python
class Section:
    def __init__(self, name):
        self.name = name

    def __enter__(self):
        print(f"--- BEGIN {self.name} ---")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print(f"--- END {self.name} ---")

with Section("Calm"):
    print("everything is fine")

try:
    with Section("Risky"):
        print("about to fail")
        raise ValueError("bad data")
        print("never reached")
except ValueError as e:
    print("caught outside:", e)     # the END line printed anyway — that's the guarantee
```

```python
# The short way: a generator with ONE yield, decorated. Unit 18 meets Unit 20.
from contextlib import contextmanager

@contextmanager
def section(name):
    print(f"--- BEGIN {name} ---")
    yield
    print(f"--- END {name} ---")

with section("Quick"):
    print("inside the block")
```

## 20.5 · Your build — the report runner (the Advanced finale)

```python
def logged(func):
    def inner(*args, **kwargs):
        print(f"[log] calling {func.__name__}")
        result = func(*args, **kwargs)
        print(f"[log] {func.__name__} done")
        return result
    return inner


class Section:
    def __init__(self, name):
        self.name = name

    def __enter__(self):
        print(f"--- BEGIN {self.name} ---")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print(f"--- END {self.name} ---")


@logged
def sales_summary(items):
    total = 0
    for price in items.values():
        total = total + price
    print(f"  {len(items)} items, total {total:.2f}")


@logged
def top_seller(items):
    if not items:
        raise ValueError("no sales to rank")
    best = ""
    best_price = 0
    for name, price in items.items():
        if price > best_price:
            best = name
            best_price = price
    print(f"  best: {best} at {best_price:.2f}")


monday = {"tea": 12.5, "cake": 45.25, "dosa": 30.0}
tuesday = {}

with Section("Monday"):
    sales_summary(monday)
    top_seller(monday)

try:
    with Section("Tuesday"):
        top_seller(tuesday)
except ValueError as e:
    print("report skipped:", e)

# One [log] ... done line is missing from Tuesday. Work out why before reading on:
# the raise happened INSIDE func(...), so the wrapper's next line never ran.
# The END line printed anyway, because __exit__ always does.
```
