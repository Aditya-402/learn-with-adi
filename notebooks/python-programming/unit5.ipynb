## 5.2 · Changing a list — append, insert, remove, pop (mutation, not reassignment)

```python
guests = ["Ada", "Grace"]
guests.append("Alan")
print(guests)
guests.insert(1, "Edsger")
print(guests)
gone = guests.pop(0)
print("Left early:", gone)
print(guests)
```

## 5.4 · Loops meet lists — `for item in my_list`, totals and counts

```python
marks = [72, 85, 90, 66, 88]
print("Average:", sum(marks) / len(marks))
high = 0
for m in marks:
    if m >= 80:
        high = high + 1
print("Scored 80+:", high)
```
