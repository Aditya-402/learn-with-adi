## 12.1 · The pattern behind the loop — read it aloud: "a list of n-times-two, for each n in numbers"

```python
numbers = [3, 8, 12]
words = ["tea", "samosa", "idli"]
names = ["ada", "bob"]

doubled = [n * 2 for n in numbers]
lengths = [len(w) for w in words]
shouted = [name.upper() for name in names]

print(doubled)
print(lengths)
print(shouted)
```

```python
# P1 — rewrite this loop as ONE comprehension line plus the print.
# target -> [20, 45, 60]
prices = [40, 90, 120]
sale = []
for p in prices:
    sale.append(p // 2)
print(sale)
```

## 12.2 · The condition clause — a gate at the end; failures are skipped entirely

```python
scores = [34, 67, 88, 45, 91]
passed = [s for s in scores if s >= 50]
print(passed)
print(len(passed), "out of", len(scores), "passed")
```

## 12.3 · Dict & set comprehensions — the braces family (sorted() before printing a set)

```python
# Unit 6's word counter, revisited in comprehension spirit.
# Honest note: the Unit 6 loop with .get() is still the better counter for big lists —
# .count() re-scans the whole list once per distinct word.
text = "the tea the chai the tea"
words = text.split()
counts = {w: words.count(w) for w in sorted(set(words))}
print(counts)
```

## 12.4 · The judgment — the two-question test: pure transform/filter? one honest line?

```python
scores = [34, 67, 88]

labels = []
for s in scores:
    if s >= 50:
        labels.append("pass")
    else:
        labels.append("fail")
print(labels)

labels2 = ["pass" if s >= 50 else "fail" for s in scores]
print(labels2)
```

## 12.5 · Your build — the marks report (the Beginner checkpoint)

New tool: `zip(names, scores)` closes two parallel lists like a zipper — loop over it with two-name unpacking, exactly like `.items()`.

```python
# Stage 1 — zip the lists.
names = ["Ada", "Bob", "Cara", "Dev", "Esha"]
scores = [91, 45, 78, 85, 62]
for name, score in zip(names, scores):
    print(name, score)
```

```python
# Stage 3 — the full report. Build: a name→grade dict via dict comprehension over zip,
# the average with :.2f, passed scores (>= 50), distinction names (>= 85),
# and aligned report lines with f"{name:<6}{score:>3}  {grades[name]}".
def grade(score):
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 50:
        return "C"
    else:
        return "F"

names = ["Ada", "Bob", "Cara", "Dev", "Esha"]
scores = [91, 45, 78, 85, 62]
```
