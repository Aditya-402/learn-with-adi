## 2.2 · Numbers & arithmetic — the seven operators

## 2.4 · f-strings — sentences without the scaffolding

```python
points = 95
price = 1234.5
print(f"Score: {points}!")
print(f"Price: {price:.2f} rupees")
```

```python
# Why money always gets :.2f — look closely at the first line.
total = 0.1 + 0.2
print("Total:", total)
print(f"Total: {total:.2f}")
```

## 2.5 · Your build — the tip calculator

```python
bill = int(input("Bill amount? "))
tip_percent = int(input("Tip percent? "))
people = int(input("How many people? "))
tip = bill * tip_percent / 100
total = bill + tip
each = total / people
print(f"Tip: {tip:.2f}")
print(f"Total: {total:.2f}")
print(f"Each pays: {each:.2f}")
```
