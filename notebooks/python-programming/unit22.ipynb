## 22.1 · The DataFrame — a table with names

A DataFrame is a dict of NumPy arrays that all agree on their row labels. Build one from a dict of lists: the keys become the column names.

```python
import pandas as pd

data = {
    "name":  ["Asha", "Ben", "Chitra", "Dev", "Esha", "Farid"],
    "class": ["A", "B", "A", "B", "A", "B"],
    "score": [72, 45, 88, 61, 54, 90],
}
df = pd.DataFrame(data)

print("shape:", df.shape)
print("columns:", list(df.columns))
print()
print(df.dtypes)
print()
print(df.head())
```

## 22.2 · Selecting — one column, several columns, some rows

One pair of brackets gives a Series. Two pairs gives a smaller DataFrame. A condition inside the brackets keeps whole rows.

```python
import pandas as pd

df = pd.DataFrame({
    "name":  ["Asha", "Ben", "Chitra", "Dev", "Esha", "Farid"],
    "class": ["A", "B", "A", "B", "A", "B"],
    "score": [72, 45, 88, 61, 54, 90],
})

print("--- one column (a Series) ---")
print(df["score"])
print()
print("--- two columns (a DataFrame) ---")
print(df[["name", "score"]])
print()
print("--- the rows that passed ---")
print(df[df["score"] >= 50])
```

## 22.4 · Missing values — the thing that makes real data real

NaN means *we don't know* — not 0, not "". Count the holes with `isna().sum()`, then choose between `dropna()` and `fillna()` and be ready to defend the choice.

```python
import pandas as pd

csv_text = """name,class,score
Asha,A,72
Ben,B,
Chitra,A,88
Dev,B,61
Esha,A,
Farid,B,90
"""
with open("gappy.csv", "w") as f:
    f.write(csv_text)

df = pd.read_csv("gappy.csv")

print(df)
print()
print("--- where are the holes? ---")
print(df.isna().sum())
print()
print("score column dtype:", df["score"].dtype)
print(f"rows: {len(df)}   real scores: {df['score'].count()}")
print(f"mean, ignoring gaps : {df['score'].mean():.2f}")
print(f"mean, gaps as zero  : {df['score'].sum() / len(df):.2f}")
print(f"mean, gaps dropped  : {df.dropna()['score'].mean():.2f}")
```

## 22.5 · Sorting, new columns & groupby

`sort_values` reorders whole rows. `df["new"] = ...` creates a column. `groupby` is split–apply–combine — the one-line answer to a question that took a dict of lists in Unit 6.

```python
import pandas as pd

df = pd.DataFrame({
    "name":  ["Asha", "Ben", "Chitra", "Dev", "Esha", "Farid"],
    "class": ["A", "B", "A", "B", "A", "B"],
    "score": [72, 45, 88, 61, 54, 90],
})

df["passed"] = df["score"] >= 50

ranked = df.sort_values("score", ascending=False)
print(ranked)
print()
print("Passed:", df["passed"].sum(), "of", len(df))
print()
for cls, m in df.groupby("class")["score"].mean().items():
    print(f"Class {cls}: {m:.2f}")
```

```python
# P3 — one readable line per class: how many students, and their average.
# target ->  Class A: 3 students, average 71.33 / Class B: 3 students, average 65.33
```
