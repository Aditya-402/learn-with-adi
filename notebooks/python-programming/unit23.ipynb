## 23.2 · Four charts, and when to use which

| your question | chart | the call |
|---|---|---|
| How did one thing change over time? | line | `plt.plot(x, y)` |
| Which category is biggest? | bar | `plt.bar(names, values)` |
| Do these two numbers move together? | scatter | `plt.scatter(x, y)` |
| What does one column's spread look like? | histogram | `plt.hist(values, bins=...)` |

The histogram is the odd one: it takes a **single** list and counts the values into
ranges (bins) itself — you never supply the bar heights.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
month_no = [1, 2, 3, 4]
sales = [120, 95, 143, 88]
hours = [1, 2, 2, 3, 4, 4, 5, 6]
marks = [52, 55, 61, 64, 70, 68, 79, 88]
all_marks = [41, 55, 58, 62, 63, 65, 66, 68, 71, 72, 74, 77, 81, 90]

plt.figure()
plt.plot(month_no, sales)
plt.title("Sales over the year so far")
plt.xlabel("Month number")
plt.ylabel("Sales (thousands)")

plt.figure()
plt.bar(months, sales)
plt.title("Sales by month")
plt.xlabel("Month")
plt.ylabel("Sales (thousands)")

plt.figure()
plt.scatter(hours, marks)
plt.title("Study hours against marks")
plt.xlabel("Hours studied per week")
plt.ylabel("Mark")

plt.figure()
plt.hist(all_marks, bins=5)
plt.title("How the class marks are spread")
plt.xlabel("Mark")
plt.ylabel("Number of students")
```

## 23.3 · Making it readable — size, colour and a legend

## 23.4 · Plotting straight from pandas

`df.plot(kind="bar", x="city", y="sales")` — and a groupby result plots itself, index and all.

```python
import pandas as pd
import matplotlib.pyplot as plt

sales = pd.DataFrame({
    "city": ["Chennai", "Pune", "Delhi", "Kochi"],
    "sales": [120, 95, 143, 88],
})
print(sales)

sales.plot(kind="bar", x="city", y="sales", color="#5b4b9e", legend=False)
plt.title("Sales by city")
plt.xlabel("City")
plt.ylabel("Sales (thousands)")
```

```python
# P2/P3 — group, plot the Series, then print the numbers behind the picture.
# Expected printout:
# Chennai: 130.0
# Delhi: 150.0
# Pune: 100.0
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "city": ["Chennai", "Chennai", "Pune", "Pune", "Delhi", "Delhi"],
    "day": ["Mon", "Tue", "Mon", "Tue", "Mon", "Tue"],
    "sales": [120, 140, 95, 105, 143, 157],
})

by_city = df.groupby("city")["sales"].mean()
```

## 23.5 · seaborn — the section that only runs here

On the unit page these three are read-only listings, because the in-page Python cannot
fetch seaborn. Colab has it installed. Run the table cell first, then the three below it.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

marks = pd.DataFrame({
    "name": ["Asha", "Ravi", "Meera", "Asha", "Ravi", "Meera", "Asha", "Ravi", "Meera"],
    "subject": ["Maths", "Maths", "Maths", "Science", "Science", "Science",
                "English", "English", "English"],
    "marks": [78, 66, 90, 88, 71, 81, 72, 84, 66],
})

study = pd.DataFrame({
    "hours": [1, 2, 3, 4, 5, 6, 2, 3, 4, 5, 6, 7],
    "mark": [45, 52, 58, 66, 70, 75, 51, 60, 64, 72, 78, 85],
    "subject": ["Maths"] * 6 + ["Science"] * 6,
})
marks
```

```python
# Listing 1 — sns.barplot: seaborn does the groupby and the mean itself,
# and the thin line on each bar shows how much the group varied.
sns.barplot(data=marks, x="subject", y="marks")
plt.title("Average mark by subject")
plt.xlabel("Subject")
plt.ylabel("Mark")
```

```python
# Listing 2 — sns.histplot: the shape of one column, plus the smoothed guess (kde).
sns.histplot(data=marks, x="marks", bins=6, kde=True)
plt.title("Spread of all marks")
plt.xlabel("Mark")
plt.ylabel("Number of results")
```

```python
# Listing 3 — sns.scatterplot with hue: a third column, coloured and legended in one word.
# Add style="subject" so the marker shapes carry the same information as the colours.
sns.scatterplot(data=study, x="hours", y="mark", hue="subject", style="subject")
plt.title("Hours against marks, split by subject")
plt.xlabel("Hours studied per week")
plt.ylabel("Mark")
```

## 23.6 · Your build — a mini data study

Four beats, always in this order: **load** the data, **compute** the numbers,
**draw** the picture, **write** what you found. The last beat is the only one with
no library and the only one anybody else will read.

```python
import pandas as pd
import matplotlib.pyplot as plt

marks = pd.DataFrame({
    "name": ["Asha", "Ravi", "Meera", "Asha", "Ravi", "Meera", "Asha", "Ravi", "Meera"],
    "subject": ["Maths", "Maths", "Maths", "Science", "Science", "Science",
                "English", "English", "English"],
    "marks": [78, 66, 90, 88, 71, 81, 72, 84, 66],
})
print(marks)
print()

by_subject = marks.groupby("subject")["marks"].mean()
print(by_subject)

by_subject.plot(kind="bar", color="#5b4b9e", figsize=(6, 4))
plt.title("Average mark by subject")
plt.xlabel("Subject")
plt.ylabel("Average mark")
```
